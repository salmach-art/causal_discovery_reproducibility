# 06 — Expert knowledge as structural constraints

Asia and Elderly were evaluated with and without required/forbidden edges.



```python


import pandas as pd

# Illustrative excerpt of the expert-knowledge processing.
# The complete constraint-handling and structure-learning
# implementation is part of a separate research project.

# LOAD EXPERT CONSTRAINTS

constraints = pd.read_csv(
    "../configs/expert_constraints.csv"
)

for cbn_name in ["Asia", "Elderly"]:

    cbn_constraints = constraints[
        constraints["CBN"] == cbn_name
    ]

    required_edges = cbn_constraints[
        cbn_constraints["ConstraintType"] == "required"
    ][["Source", "Target"]].values.tolist()

    forbidden_edges = cbn_constraints[
        cbn_constraints["ConstraintType"] == "forbidden"
    ][["Source", "Target"]].values.tolist()

    print(f"{cbn_name} — Required:", required_edges)
    print(f"{cbn_name} — Forbidden:", forbidden_edges)

    # ...
    # Constraints are subsequently provided to the
    # structure-learning procedure.
```
