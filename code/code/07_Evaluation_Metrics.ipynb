# 07 — Evaluation metrics

Metrics: SHD, nSHD, undirected-edge precision/recall/F1 and directed-edge
precision/recall/F1.

For the directed evaluation, a reverse arc is counted as a wrong orientation.
Partially oriented edges are not counted as true-positive directed edges.

```python
def directed_metrics(reference_edges, predicted_edges):
    ref, pred = set(reference_edges), set(predicted_edges)
    tp = len(ref & pred)
    wrong = sum(1 for a,b in pred if (b,a) in ref and (a,b) not in ref)
    extra = sum(1 for e in pred if e not in ref and (e[1],e[0]) not in ref)
    missing = sum(1 for e in ref if e not in pred and (e[1],e[0]) not in pred)
    shd = extra + missing + wrong
    nshd = shd / len(ref) if ref else float("nan")
    precision = tp / len(pred) if pred else 0
    recall = tp / len(ref) if ref else 0
    f1 = 2*precision*recall/(precision+recall) if precision+recall else 0
    return {"TP":tp,"Extra":extra,"Missing":missing,
            "WrongOrientation":wrong,"SHD":shd,"nSHD":nshd,
            "EdgePrecision":precision,"EdgeRecall":recall,"EdgeF1":f1}
```
