# 02 — Synthetic / semi-synthetic data generation

The study generated experimental data from expert-defined CBNs.

Asia and Elderly: semi-synthetic generation based on the original observational
data/reference parameters.

Suicide: fully synthetic generation from the expert-defined CBN and CPTs.

Sample sizes: 500, 1,000, 5,000, 10,000 and 50,000.


```python
import numpy as np
import pandas as pd


def generate_semi_synthetic_data(
    original_data,
    n_samples,
    output_file,
    random_state=42
):
    """
    Generate semi-synthetic data by bootstrap sampling
    from the original observational dataset.
    """

    synthetic_data = original_data.sample(
        n=n_samples,
        replace=True,
        random_state=random_state
    ).reset_index(drop=True)

    synthetic_data.to_csv(
        output_file,
        index=False
    )

    return synthetic_data


# ASIA
# ============================================================

AsiaSynthetic = generate_semi_synthetic_data(
    original_data=AsiaData,
    n_samples=10000,
    output_file="Asia_SemiSynthetic_10000.csv"
)



# ELDERLY
# ============================================================

ElderlySynthetic = generate_semi_synthetic_data(
    original_data=ElderlyData,
    n_samples=10000,
    output_file="Elderly_SemiSynthetic_10000.csv"
)

```
