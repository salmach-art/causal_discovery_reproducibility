# 01 — CBN structures and CPTs

This notebook documents the reference causal Bayesian networks used in the study.

This is a part of documented implementation of the procedures described in the manuscript.


```python
import pyAgrum as gum

# --------------------------------------------------
# Create the Asia Causal Bayesian Network
# --------------------------------------------------

asia_bn = gum.BayesNet("Asia")

# Add variables
asia_bn.add(gum.LabelizedVariable("VisitAsia", "Visit Asia", 2))
asia_bn.add(gum.LabelizedVariable("Smoking", "Smoking", 2))
asia_bn.add(gum.LabelizedVariable("Tuberculosis", "Tuberculosis", 2))
asia_bn.add(gum.LabelizedVariable("LungCancer", "Lung Cancer", 2))
asia_bn.add(gum.LabelizedVariable("Bronchitis", "Bronchitis", 2))
asia_bn.add(gum.LabelizedVariable("TubOrLung", "Tuberculosis or Lung Cancer", 2))
asia_bn.add(gum.LabelizedVariable("Dyspnea", "Dyspnea", 2))
asia_bn.add(gum.LabelizedVariable("Xray", "X-ray", 2))

asia_bn.addArc("VisitAsia", "Tuberculosis")
asia_bn.addArc("Smoking", "LungCancer")
asia_bn.addArc("Smoking", "Bronchitis")
asia_bn.addArc("Tuberculosis", "TubOrLung")
asia_bn.addArc("LungCancer", "TubOrLung")
asia_bn.addArc("TubOrLung", "Dyspnea")
asia_bn.addArc("Bronchitis", "Dyspnea")
asia_bn.addArc("TubOrLung", "Xray")

# ---------------------

cpt_VisitAsia = asia_bn.cpt("VisitAsia")
cpt_Tuberculosis = asia_bn.cpt("Tuberculosis")
cpt_Smoking = asia_bn.cpt("Smoking")
cpt_LungCancer = asia_bn.cpt("LungCancer")
cpt_Bronchitis = asia_bn.cpt("Bronchitis")
cpt_TubOrLung = asia_bn.cpt("TubOrLung")
cpt_Dyspnea = asia_bn.cpt("Dyspnea")
cpt_Xray = asia_bn.cpt("Xray")

# ---------------------

cpt_VisitAsia.fillWith([0.99, 0.01])
cpt_Smoking.fillWith([0.5, 0.5])

cpt_Tuberculosis[{"VisitAsia": 0}] = [0.99, 0.01]
cpt_Tuberculosis[{"VisitAsia": 1}] = [0.95, 0.05]

cpt_TubOrLung[{"Tuberculosis": 0, "LungCancer": 0}] = [1.0, 0.0]
cpt_TubOrLung[{"Tuberculosis": 0, "LungCancer": 1}] = [0.0, 1.0]
cpt_TubOrLung[{"Tuberculosis": 1, "LungCancer": 0}] = [0.0, 1.0]
cpt_TubOrLung[{"Tuberculosis": 1, "LungCancer": 1}] = [0.0, 1.0]

cpt_Dyspnea[{"TubOrLung": 0, "Bronchitis": 0}] = [0.90, 0.10]
cpt_Dyspnea[{"TubOrLung": 0, "Bronchitis": 1}] = [0.30, 0.70]
cpt_Dyspnea[{"TubOrLung": 1, "Bronchitis": 0}] = [0.20, 0.80]
cpt_Dyspnea[{"TubOrLung": 1, "Bronchitis": 1}] = [0.05, 0.95]

cpt_Xray[{"TubOrLung": 0}] = [0.95, 0.05]
cpt_Xray[{"TubOrLung": 1}] = [0.10, 0.90]


# ----------------------

print(asia_bn)

for node in asia_bn.nodes():
    print("\n-----------------------------")
    print("CPT:", node)
    print(asia_bn.cpt(node))
```
